Datastucture

```python

#List

fruits=["apple","banana","orange"]
print(fruits)
print(fruits[1]) # show 2 nd index value , index start from 0
fruits.append("grapes") # add new value to list 
print(fruits)
fruits[1]="kiwi" # change value of list 
print(fruits)

```

```python
#Tuple
colours=("red","green","blue")
print(colours)
print(colours[1]) # show 2nd index value, index start from 0
```

##Control flow

If...elif..else

```python
age=int(input("Enter your age : "))
if age>=18:
    print("you can vote")
else:
    print("You cant vote")
```

```python
age=str(input("Enter your age : "))
if age=="eighteen":
    print("you can vote")
else:
    print("You cant vote")
```

```python
money=int(input("Enter the amount:"))
if money>=100:
    print("Buy briyani")
elif money<=100 and money>=80:
    print("full meals")
elif money<=80 and money>=40:
    print("Buy curd rice")
elif money<40 and money>=10:
    print("buy vada and tea")
else:
    print("drink water and go home")
```

##Nested IF

```python
money=int(input("Enter the amount"))
shop=str(input("shop open or close"))

if money>=50:
    if shop== "open":
        print("Order pongal")
    else:
        print("search for another hotel")
else:
    print("go home")
```

##Error handling ( showing msg instead code crashing)

```python
try:                                   # write the code were error is
    result=10/2
except ZeroDivisionError:              # shows when error present
    print("this is an error")
else:                                  # it no error show the rsults
    print("NO error",result)
finally:
    print("this allway runs")          # always runs
```

## File handling

open a new or overwrites teh file

```python
with open("myfile.txt","w") as file:
    file.write("this is my file .\n")
    file.write("my name is allwyn")
print("file is written succes")    
```

Readin the file

```python
with open("myfile.txt","r") as file:
    content=file.read()
print(content)

```

appending a file or adding a new file

```python
with open("myfile.txt", "a") as file:
    file.write("\n this my new file")
print("Text added")
```
